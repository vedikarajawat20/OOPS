1. What is Object-Oriented Programming (OOP)?


Object-Oriented Programming (OOP) is a programming paradigm that organizes code into objects that represent real-world entities or concepts. Encapsulation: This is the bundling of data (attributes) and methods that operate on the data into a single unit or class. Abstraction: Abstraction involves hiding the complex implementation details of a system and exposing only the essential parts to the user. It helps to reduce complexity and allows the programmer to work with high-level concepts instead of worrying about the underlying details.

Inheritance: Inheritance allows a new class (subclass) to inherit attributes and methods from an existing class (parent class). This promotes code reuse and helps model hierarchical relationships between objects (e.g., a Dog class might inherit from an Animal class).

Polymorphism: Polymorphism enables objects of different classes to be treated as objects of a common superclass. It allows methods to have different implementations based on the object calling them (e.g., a method speak() in both Dog and Cat classes can perform different tasks, but can be called in the same way).

2. What is a class in OOP?


In Object-Oriented Programming (OOP), a class is a blueprint or template for creating objects (instances). It defines the structure and behavior that the objects created from the class will have. A class typically includes:

Attributes (Properties): These are variables that store the state or data of an object. They represent the characteristics of the object.

Methods (Functions): These are functions that define the behavior or actions the object can perform. They operate on the attributes of the object.

In simple terms, you can think of a class as a blueprint for an object, and an object is an instance of that class. The class defines what an object will have (attributes) and what it can do (methods).

Example of a Class in Python:

```python
class Dog:
    # Constructor method (initializes an object)
    def __init__(self, name, breed):
        self.name = name  # Attribute
        self.breed = breed  # Attribute

    # Method to describe the dog's behavior
    def bark(self):
        return f"{self.name} says Woof!"

# Creating an instance of the Dog class (an object)
my_dog = Dog("Buddy", "Golden Retriever")

# Accessing attributes and calling methods of the object
print(my_dog.name)   # Output: Buddy
print(my_dog.bark())  # Output: Buddy says Woof!
```

13. What is a property decorator in Python?


In Python, the @property decorator is used to define a method as a getter for an attribute, allowing you to access it like a regular attribute without explicitly calling a method. This allows for encapsulation of the class's internal data and provides a way to define custom behavior for getting or setting the value of an attribute.

What Does the @property Decorator Do?
It allows you to define a method that behaves like an attribute.

It enables you to calculate or modify the value of an attribute dynamically when it is accessed.

It hides the internal representation of an attribute, allowing you to control how it is accessed and modified (i.e., getter and setter methods).

Using @property:
Getter: The @property decorator is applied to a method that retrieves (gets) the value of an attribute.

Setter: The @<attribute>.setter decorator is used to define a setter method for the property, which allows you to modify the value of the attribute in a controlled way.

Example of Using @property

```python
class Circle:
    def __init__(self, radius):
        self._radius = radius

    @property
    def radius(self):
        """Getter method for the radius"""
        return self._radius

    @radius.setter
    def radius(self, value):
        """Setter method for the radius"""
        if value < 0:
            raise ValueError("Radius cannot be negative!")
        self._radius = value

    @property
    def area(self):
        """Calculated property for the area"""
        return 3.14 * (self._radius ** 2)

# Creating an object
circle = Circle(5)

# Accessing the radius using the getter method
print(circle.radius)  # Output: 5

# Setting the radius using the setter method
circle.radius = 10
print(circle.radius)  # Output: 10

# Accessing the calculated area (no need for method call)
print(circle.area)  # Output: 314.0

# Trying to set a negative radius will raise an exception
# circle.radius = -3  # Raises ValueError: Radius cannot be negative!
```

15.What is an abstract class in Python?


In Python, an abstract class is a class that cannot be instantiated directly and serves as a blueprint for other classes. It is used to define common methods and properties that must be implemented by any subclass. An abstract class can contain abstract methods, which are methods that have no implementation in the abstract class but must be implemented by any subclass that inherits from it.

Key Points About Abstract Classes:
Cannot Be Instantiated: An abstract class cannot be instantiated directly. It can only be used as a base class for other classes that provide implementations for its abstract methods.

Abstract Methods: These are methods defined in an abstract class without an implementation. Subclasses that inherit from this abstract class are required to implement these abstract methods.

Use of abc Module: To define abstract classes and methods in Python, we use the abc (Abstract Base Class) module, which provides the tools to create and manage abstract classes.

How to Define an Abstract Class:
Import ABC and abstractmethod: You need to import ABC (the base class for defining abstract classes) and abstractmethod (used to define abstract methods) from the abc module.

Inheritance: Your abstract class should inherit from ABC.

Define Abstract Methods: Use the @abstractmethod decorator to mark methods that must be implemented in subclasses.

Example of Abstract Class in Python:


```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

    @abstractmethod
    def move(self):
        pass

# This will raise an error because Animal is abstract and cannot be instantiated
# animal = Animal()  # TypeError: Can't instantiate abstract class Animal with abstract methods sound, move

class Dog(Animal):
    def sound(self):
        return "Woof!"

    def move(self):
        return "The dog runs!"

class Bird(Animal):
    def sound(self):
        return "Chirp!"

    def move(self):
        return "The bird flies!"

# Now we can create instances of Dog and Bird
dog = Dog()
bird = Bird()

print(dog.sound())  # Output: Woof!
print(dog.move())   # Output: The dog runs!

print(bird.sound())  # Output: Chirp!
print(bird.move())   # Output: The bird flies!
```

17. What is the difference between a class variable and an instance variable?


In Object-Oriented Programming (OOP), class variables and instance variables are two types of variables that are used to store data in a class. While they may seem similar, they serve different purposes and have distinct behaviors. Here’s a detailed comparison of the two:

1. Class Variable:
Definition: A class variable is a variable that is shared by all instances of a class. It is declared within the class, but outside any instance methods (including the constructor).

Access: Class variables can be accessed using the class name as well as via any instance of the class.

```python
class Dog:
    species = "Canis familiaris"  # Class variable

    def __init__(self, name):
        self.name = name  # Instance variable

# All dogs share the same species
dog1 = Dog("Buddy")
dog2 = Dog("Charlie")

print(dog1.species)  # Output: Canis familiaris
print(dog2.species)  # Output: Canis familiaris

# Changing the class variable will affect all instances
Dog.species = "Canis lupus"
print(dog1.species)  # Output: Canis lupus
print(dog2.species)  # Output: Canis lupus
```

18. What is multiple inheritance in Python?


Multiple inheritance in Python refers to a feature where a class can inherit attributes and methods from more than one parent class. This allows a derived class to combine functionalities from multiple base classes, making it a powerful tool for code reuse and flexible design.

Key Points About Multiple Inheritance:
Inheriting from Multiple Classes: A subclass can inherit from more than one class, thus gaining access to the methods and attributes of all its parent classes.

Method Resolution Order (MRO): When a method is called on an object, Python needs to determine which method to execute. The method search follows a specific order called the Method Resolution Order (MRO). This ensures that Python doesn't get confused when multiple parent classes have methods with the same name. MRO can be inspected using the mro() method.

Diamond Problem: In multiple inheritance, a class can inherit from two classes that have a common ancestor. This can sometimes lead to ambiguity (known as the diamond problem), but Python’s MRO system uses C3 Linearization to solve this by ensuring a clear order of method resolution.

Example of Multiple Inheritance:


```python
class Animal:
    def speak(self):
        print("Animal speaking")

class Mammal(Animal):
    def walk(self):
        print("Mammal walking")

class Bird(Animal):
    def fly(self):
        print("Bird flying")

# Dog class inherits from both Mammal and Bird
class Dog(Mammal, Bird):
    def bark(self):
        print("Dog barking")

# Creating an instance of Dog
dog = Dog()

dog.speak()  # Inherited from Animal class
dog.walk()   # Inherited from Mammal class
dog.fly()    # Inherited from Bird class
dog.bark()   # Defined in Dog class
```

23. How does polymorphism work in Python with inheritance?


In Python, polymorphism refers to the ability of different classes to provide a unique implementation of a method that is defined in a common interface (typically a parent or base class). Inheritance is one of the primary ways polymorphism works in Python. It allows a subclass to override or extend the methods of its parent class, providing its own specific implementation while maintaining the same method name.

Polymorphism in Python works through method overriding and dynamic method resolution. The key point is that the type of the object (the actual class) determines which method gets called, rather than the type of the reference variable.

Key Concepts of Polymorphism in Python with Inheritance:
Method Overriding: A subclass can provide its own implementation of a method that is already defined in its parent class. When an object of the subclass calls the method, the subclass version of the method is invoked.

Dynamic Method Resolution: Python uses dynamic typing and resolves method calls at runtime. This means the method that gets executed is determined by the actual class of the object, not the type of the reference variable.Same Method Name, Different Behavior: The ability for different classes to define methods with the same name but behave differently (based on the class the object belongs to) is the essence of polymorphism.

Example of Polymorphism in Python with Inheritance:
Let's look at an example where polymorphism works with inheritance:


```python
# Parent class (Base class)
class Animal:
    def make_sound(self):
        print("Animal makes a sound")

# Child class 1
class Dog(Animal):
    def make_sound(self):
        print("Dog barks")

# Child class 2
class Cat(Animal):
    def make_sound(self):
        print("Cat meows")

# Creating objects of different classes
animals = [Animal(), Dog(), Cat()]

# Polymorphism in action: calling the same method on different objects
for animal in animals:
    animal.make_sound()
```

To enable method chaining in Python, each method in the chain must return the object (self) or another object that allows subsequent method calls. By returning self, we ensure that the object itself can be used for further method calls.

Example of Method Chaining in Python:

```python
class Calculator:
    def __init__(self, value=0):
        self.value = value

    def add(self, number):
        self.value += number
        return self  # Return the object itself to allow chaining

    def subtract(self, number):
        self.value -= number
        return self  # Return the object itself to allow chaining

    def multiply(self, number):
        self.value *= number
        return self  # Return the object itself to allow chaining

    def get_value(self):
        return self.value


# Method chaining in action
calc = Calculator()
result = calc.add(5).subtract(2).multiply(3).get_value()

print(result)  # Output: 9
```

```python
class Student:
    def __init__(self, name, grades):
        # Initialize the attributes: name and grades
        self.name = name
        self.grades = grades

    # Method to calculate the average of the grades
    def average_grade(self):
        if len(self.grades) == 0:
            return 0  # Prevent division by zero if there are no grades
        return sum(self.grades) / len(self.grades)

# Create an instance of the Student class
student = Student("Alice", [90, 80, 85, 88, 92])

# Call the average_grade method
average = student.average_grade()
print(f"{student.name}'s average grade is: {average:.2f}")  # Output: Alice's average grade is: 87.00
```

14. Create a class Employee with a method calculate_salary() that computes the salary based on hours worked
and hourly rate. Create a derived class Manager that adds a bonus to the salary.

```python
# Base class Employee
class Employee:
    def __init__(self, name, hours_worked, hourly_rate):
        self.name = name
        self.hours_worked = hours_worked
        self.hourly_rate = hourly_rate

    def calculate_salary(self):
        return self.hours_worked * self.hourly_rate


# Derived class Manager (inherits from Employee)
class Manager(Employee):
    def __init__(self, name, hours_worked, hourly_rate, bonus):
        # Initialize parent class (Employee)
        super().__init__(name, hours_worked, hourly_rate)
        self.bonus = bonus

    def calculate_salary(self):
        # Calculate salary using the base method and add the bonus
        base_salary = super().calculate_salary()
        return base_salary + self.bonus


# Example Usage:
# Creating an Employee object
employee = Employee("John Doe", 160, 25)
print(f"Employee Salary: ${employee.calculate_salary()}")

# Creating a Manager object
manager = Manager("Jane Smith", 160, 30, 5000)
print(f"Manager Salary: ${manager.calculate_salary()}")
```
